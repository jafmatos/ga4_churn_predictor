## Configuração da Página

O Streamlit disponibiliza um método que configura os metadados da página. Este método é o set_page_config. Os principais parâmetros que ele define são: page_title; page_icon; layout; initial_sidebar_state. O page_title define o título da página. O page_icon define o favicon. O parâmetro layout define o estilo geral da página, se quiesermos que o conteúdo ocupe toda a largura da página, atribuímos "wide", e se quiesermos que o conteúdo ocupe menos espaço, podemos atribuir "centered" que centralizará o conteúdo para que tenha uma largura menor. O initial_sidebar_state definirá o estado inicial da barra lateral da página. Podendo receber "expanded" para expandida, "collapsed" para fechada, "auto" para manter o comportamento natural e "locked" para manter a barra aberta permanentemente desde o início. 

````python
import streamlit

streamlit.set_page_config(
    page_title="Preditor de Churn GA4",
    page_icon="🧠",
    layout="centered",
    initial_sidebar_state="expanded"
)
````

## Textos e Tipografia

Antes de tudo, eis o método mais simples do Streamlit que carrega um texto na página. O método write recebe como primeiro parâmetro o conteúdo que será carregado na página. Esse conteúdo pode conter tags HTML. Quando quisermos que o HTML seja interpretado, basta utilizar o parâmetro unsafe_allow_html que define se o HTML será interpretado ou não. Permitir a interpretação de código de terceiros pode ser arriscado. Portanto, utilize cuidadosamente.

````python
streamlit.write(
    "Este modelo utiliza dados comportamentais do <strong>Google Analytics 4</strong> para calcular o risco de evasão.",
    unsafe_allow_html=True,
)
````

### Títulos e Legendas

O Streamlit tem alguns métodos para títulos e legendas. São eles: title, header, subheader e caption. Todos eles recebem como primeiro parâmetro ou como parâmetro body uma string com o conteúdo do título ou legenda.

````python
streamlit.title("Painel de Previsão de Churn GA4",)
streamlit.header(body="Análise Preditiva de Clientes")
streamlit.subheader(body="Indicadores do Modelo")
streamlit.caption("Versão do modelo: v1.0.0 | Último retreino: Outubro/2026")
````

![Figura 2](assets/images/figura_2.png)

Todos os métodos de títulos e legenda podem receber um parâmetro text_alignment, que pode receber "center" para alinhar o texto ao centro, "left" para alinhar o texto à esquerda, "right" para alinhar o texto à direita e "justify" para esticar as linhas, exceto a última, para ocuparem toda a largura disponível.

````python
streamlit.title("Painel de Previsão de Churn GA4", text_alignment="center")
streamlit.header("Análise Preditiva de Clientes", text_alignment="left")
streamlit.subheader("Indicadores do Modelo", text_alignment="right")
streamlit.caption(
    "Versão do modelo: v1.0.0 | Último retreino: Outubro/2026", 
    text_alignment="justify"
)
````

![Figura 3](assets/images/figura_3.png)

Todos os métodos de títulos e legenda também podem receber um parâmetro wrap, que define se o texto sofrerá quebras de linha para caber na largura disponível. Caso o texto exceda o espaço e wrap receba False, o texto excedente será resumido em elipses.

````python
streamlit.title("O Streamlit é uma ferramenta fantástica para criação de Dashboards", wrap=True)
streamlit.header("O Streamlit é uma ferramenta fantástica para criação de Dashboards", wrap=True)
streamlit.subheader("O Streamlit é uma ferramenta fantástica para criação de Dashboards", wrap=False)
streamlit.caption(
    "O Streamlit é uma ferramenta fantástica para criação de Dashboards", 
    wrap=False
)
````

![Figura 4](assets/images/figura_4.png)

Outro parâmetro que todos os métodos de títulos e legenda podem receber é o parâmetro width. Ele pode receber "content" ou "stretch". O valor "content" fará com que o elemento que encapsula o texto tenha exatamente a largura do texto. Já o valor "stretch" fará com que o elemento que encapsula o texto ocupe toda a largura disponível. O alinhamento de texto não funciona quando o elemento de texto tem a largura do conteúdo.

````python
streamlit.title(
    "Painel de Previsão de Churn GA4", 
    text_alignment="center", 
    width="content"
)
streamlit.header(
    "Análise Preditiva de Clientes", 
    text_alignment="left", 
    width="stretch"
)
streamlit.subheader(
    "Indicadores do Modelo", 
    text_alignment="right", 
    width="stretch"
)
streamlit.caption(
    "Versão do modelo: v1.0.0 | Último retreino: Outubro/2026",
    text_alignment="justify",
    width="content",
)
````

![Figura 6](assets/images/figura_6.png)

Por fim, um dos parâmetros mais relevantes que todos os métodos de títulos e legenda podem receber é o parâmetro help. O parâmetro help receberá um texto de ajuda. Com isso, o elemento de texto terá um ícone de interrogação logo ao lado. Quando o usuário passar o mouse em cima do ícone, o texto de ajuda será exibido.

````python
streamlit.title(
    "Painel de Previsão de Churn GA4", 
    help="Algum texto de ajuda..."
)
streamlit.header(
    "Análise Preditiva de Clientes", 
    help="Algum texto de ajuda..."
)
streamlit.subheader(
    "Indicadores do Modelo", 
    help="Algum texto de ajuda..."
)
streamlit.caption(
    "Versão do modelo: v1.0.0 | Último retreino: Outubro/2026",
    help="Algum texto de ajuda..."
)
````

## Métricas

Quando vamos apresentar um protótipo integrado ao modelo, é importante que quem está avaliando e não entende da parte técnica de Machine Learning consiga compreender as métricas do modelo facilmente. Para isso, é necessário implementar uma forma intuitiva de representar as métricas. O Streamlit tem um método metric que faz exatamente isso. Dois parâmetros são obrigatórios: label e value. O parâmetro label define o nome da métrica. Já o parâmetro value define o valor. 

````python
streamlit.metric(
    label="Acurácia Global", 
    value="87.4%", 
)

streamlit.metric(
    label="AUC-ROC", 
    value="0.91", 
)

streamlit.metric(
    label="Tempo Médio de Inferência", 
    value="14 ms", 
)
````

![Figura 13](assets/images/figura_13.png)

Muitas métricas podem ter uma variação. Por exemplo quando calculamos a mesma métrica de várias formas diferentes e utilizamos a média dos resultados. Para representar isso visualmente, podemos utilizar o parâmetro delta. O Streamlit irá colorir o chip do delta de verde caso o valor inicie com um sinal de "+" ou não inicie com sinal nenhum. Caso o valor inicie com um sinal de "-", o chip será colorido de vermelho conforme o exemplo abaixo.

````python
streamlit.metric(
    label="Acurácia Global", 
    value="87.4%", 
    delta="+1.2%"
)

streamlit.metric(
    label="AUC-ROC", 
    value="0.91", 
    delta="0.03"
)

streamlit.metric(
    label="Tempo Médio de Inferência", 
    value="14 ms", 
    delta="-2 ms", 
)
````

![Figura 14](assets/images/figura_14.png)

Muitas vezes, quando a variação é negativa, não significa algo ruim. Nesse exemplo, uma variação negativa para o tempo médio de inferência é algo bom, visto que dessa forma o modelo responde mais rápido. Da forma que estava antes, a variação ficava vermelha, passando uma imagem de algo negativo. Para mudar isso, podemos passar o parâmetro delta_color que receberá "inverse" ou outra cor como "blue", "orange", etc. Quando o parâmetro delta_color recebe inverse, o chip de variação ficará verde quando a variação for negativa e vermelho quando a variação for positiva.

````python
streamlit.metric(
    label="Acurácia Global", 
    value="87.4%", 
    delta="+1.2%"
)

streamlit.metric(
    label="AUC-ROC", 
    value="0.91", 
    delta="0.03"
)

streamlit.metric(
    label="Tempo Médio de Inferência", 
    value="14 ms", 
    delta="-2 ms", 
    delta_color="inverse"
)
````

![Figura 15](assets/images/figura_15.png)

NOTA: O parâmetro delta_color, além dos valores da tabela, pode receber outros valores representando diferentes cores. Vale a pena exprimentar várias opções para fixar o mais importante.

| Parâmetro | Valores Permitidos | O que define |
| --------- | ------------------ | ------------ |
| label     |                    | Nome da métrica | 
| value     |                    | Valor principal da métrica | 
| delta     |                    | Variação do valor da métrica | 
| delta_color | "inverse", "normal", etc. | Cor da variação da métrica | 

## Dataframes e Tabelas

No Streamlit, podemos utilizar dataframes. Para incluir um dataframe visível para o usuário, usamos o método dataframe. Como primeiro parâmetro, devemos passar o dataframe com os dados a serem exibidos. O método dataframe também pode receber outros argumentos. Dentre eles, dois dos mais relevantes são: width e hide_index. O width poderá receber "content" ou "stretch" para que o dataframe ocupe ou a largura do conteúdo ou a largura total disponível, respectivamente.
````python
from ga4_churn_predictor.config import PROCESSED_DATA_DIR
import pandas

dataframe_users = pandas.read_csv(PROCESSED_DATA_DIR / "users_predicted.csv")

streamlit.dataframe(dataframe_users.sample(n=100), width="stretch", hide_index=False)
````

## Recebendo Dados e Interações do Usuário

### Dados Numéricos

Para receber dados numéricos do usuário utilizando o Streamlit, podemos fazer de duas formas. Ou com um input numérico ou com um elemento de slider. Para incluir um input numérico na página, chamamos o método number_input do Streamlit. Com o primeiro parâmetro ou parâmetro label, definimos o nome do campo. O input também poderá receber outros parâmetros como min_value, que define o valor mínimo permitido. O parâmetro max_value, que define o valor máximo permitido. O step, que define o quanto o valor tem de aumentar ou diminuir por vez. O parâmetro value define um valor inicial para o input. Além do parâmetro help que já conhecemos, que carrega um ícone ao lado do elemento e quando o usuário passa o mouse por cima do ícone, o texto de ajuda é exibido. Perceba que no caso abaixo temos um input para números com duas casas decimais, por isso, os parâmetros min_value, max_value, step e value sempre tem de ter a mesma quantidade de casas, nesse exemplo, duas.

````python
streamlit.number_input(
    "Receita Total em Dólares ($)",
    min_value=0.01,
    max_value=10000.00,
    step=0.01,
    value=1.00,
    help="Define a receita total em dólares que um usuário gerou."
)
````

O outro elemento que podemos carregar é o slider. Para isso, chamamos o método slider do Streamlit. Ele recebe exatamente os mesmos parâmetros que o number_input. Além dos outros argumentos, o método slider também pode receber o format que define uma string de formatação para o valor do slider. No exemplo abaixo, estamos formatando o valor para porcentagem sem casas decimais. Para mais informações sobre funções de formatação, visite [sprintf.js](https://github.com/alexei/sprintf.js?tab=readme-ov-file#format-specification).

````python
streamlit.slider(
    "Limiar de Probabilidade de Churn",
    min_value=1,
    max_value=100,
    step=1,
    value=50,
    format="%.0f%%",
    help="Define a probabilidade mínima para considerar um usuário como churn."
)
````

### Dados Categóricos

O Streamlit disponibiliza alguns elementos para seleção. Nós apresentaremos do mais discreto (com menos opções ou seleções) ao menos discreto (com mais opções ou seleção múltipla).

Portanto, o primeiro dos elementos é carregado pelo método radio. O primeiro parâmetro ou parâmetro label define o nome do campo. O único outro parâmetro obrigatório é options, que recebe uma lista com as opções disponíveis para seleção. Aleḿ deles, também podemos passar os parâmetros index, horizontal e width. O parâmetro index define o índice da opção marcada por padrão. O parâmetro horizontal define se os itens de seleção devem fluir de forma horizontal ou vertical. Já o parâmetro width, como de costume, pode receber "content" para que o elemento ocupe apenas a largura do conteúdo ou "stretch" para que ocupe toda a largura disponível.

````python
streamlit.radio(
    "Modelo de Previsão", 
    options=["v1", "v2"],
    index=1, 
    horizontal=True, 
    width="content"
)
````

Por último, o Streamlit disponibiliza um método multiselect que assim como os outros, recebe um primeiro parâmetro ou parâmetro label que define o nome do campo e um parâmetro options que recebe a lista de opções disponíveis para seleção. Além disso, pode receber o argumento placeholder que é um texto exibido quando o seletor está vazio. Também pode receber um parâmetro max_selections que define a quantidade máxima de itens que podem ser selecionados. Outro parâmetro que o método multiselect pode receber é select_all, que define se deve ser exibida uma opção que seleciona todos os itens automaticamente ou não. #### Session State

Antes de continuar nas funções de controle de fluxo, é conveniente aprendermos a manipular o estado a aplicação. Para melhor explicar, demonstraremos um exemplo funcional em código e depois destrincharemos o código parte por parte. Eis o código abaixo.

````python
streamlit.title("Contador")

streamlit.markdown(f"**Contagem**: {streamlit.session_state.get("contador", 0)}")

button_col_1, button_col_2, button_col_3 = streamlit.columns(3)

with button_col_1:
    if streamlit.button("Incrementar", type="primary", width="stretch"):
        if streamlit.session_state.get("contador", None) is None:
            streamlit.session_state["contador"] = 1
        else:
            streamlit.session_state["contador"] += 1

with button_col_2:
    if streamlit.button("Decrementar", type="primary", width="stretch"):
        if streamlit.session_state.get("contador", None) is None:
            streamlit.session_state["contador"] = -1
        else:
            streamlit.session_state["contador"] -= 1

with button_col_3:
    if streamlit.button("Reiniciar", type="secondary", width="stretch"):
        if streamlit.session_state.get("contador", None) is not None:
            streamlit.session_state["contador"] = 0
````

![Figura 42](assets/images/figura_42.png)

Agora, vamos destrinchar o código parte por parte, começando no elemento que exibe o valor atual do contador. Para recuperar o valor atual, acessamos a propriedade session_state do Streamlit. É através desta propriedade que gerenciaremos o estado dos dados da aplicação, caso precisemos. O método get do session_state recebe como primeiro parâmetro a chave da propriedade que queremos recuperar do estado. Como segundo parâmetro, pode receber um valor padrão caso a propriedade não esteja definida no estado. Para este exemplo, o elemento tenta exibir o valor atual do contador. Mas, caso o contador não tenha sido definido no estado ainda, exibe o número zero.

````python
streamlit.markdown(f"**Contagem**: {streamlit.session_state.get("contador", 0)}")
````

O que não se pode fazer ao tentar recuperar uma propriedade é acessar sem utilizar o método get, a menos que você tenha certeza de que naquele ponto a propriedade já exista em session_state.

````python
streamlit.markdown(f"**Contagem**: {streamlit.session_state.contador}")
````

Já recuperamos dados, agora vamos escrever ou atualizar dados. No trecho abaixo, adicionamos um botão de incrementar o contador. Quando o botão é clicado, tentamos recuperar o valor atual do contador, assumindo o valor None caso não esteja definido. Se o valor retornado for None, sabemos que o contador não foi definido e definimos o valor como 1. Para definir o valor, simplesmente acessamos a propriedade que desejamos adicionar ou atualizar e atribuímos a ela um novo valor.

````python
with button_col_1:
    if streamlit.button("Incrementar", type="primary", width="stretch"):
        if streamlit.session_state.get("contador", None) is None:
            streamlit.session_state["contador"] = 1
        else:
            streamlit.session_state["contador"] += 1
````

Também podemos fazer como no exemplo abaixo acessando diretamente a propriedade sem síntaxe de dicionário.

````python
with button_col_1:
    if streamlit.button("Incrementar", type="primary", width="stretch"):
        if streamlit.session_state.get("contador", None) is None:
            streamlit.session_state.contador = 1
        else:
            streamlit.session_state.contador += 1
````

Outra forma pela qual podemos escrever ou atualizar valores no session_state é utilizando o método update. O método update recebe um dicionário e "funde" ele com o estado da aplicação. Assim, podemos modificar mais de uma propriedade do estado na mesma chamada.

````python
with button_col_1:
    if streamlit.button("Incrementar", type="primary", width="stretch"):
        if streamlit.session_state.get("contador", None) is None:
            streamlit.session_state.update({ "contador": 1 })
        else:
            streamlit.session_state.update({ "contador": streamlit.session_state.contador + 1 })
````

Alguns métodos como keys, values e items nos trazem informações sobre o estado da aplicação. Todos eles retornam iteráveis que podem ser convertidos em listas utilizando a função list do python. O método keys retorna todos os nomes de propriedades no session_state. O método values retorna todos os valores guardados em todas as propriedades do session_state. Por fim, o método items retorna várias tuplas. Cada tupla tem como primeiro item o nome da propriedade e como segundo item o valor da propriedade.

````python
streamlit.session_state.contador = 0

print(list(streamlit.session_state.keys()))
print(list(streamlit.session_state.values()))
print(list(streamlit.session_state.items()))
````

```shell
['contador']
[0]
[('contador', 0)]
```

O terceiro botão reinicia a contagem. Podemos fazer isso de duas formas: atribuindo o valor 0 (inicial) à propriedade "contador" ou simplesmente excluindo essa propriedade do session_state. Para excluir uma propriedade do session_state, chamamos o método pop, passando como parâmetro o nome da propriedade.

````python
with button_col_3:
    if streamlit.button("Reiniciar", type="secondary", width="stretch"):
        if streamlit.session_state.get("contador", None) is not None:
            streamlit.session_state.pop("contador")
````

O último de todos os métodos mais relevantes do session_state é o método clear. O método clear simplesmente limpará todo o estado da aplicação. Não sobrará nenhum valor ou propriedade que tivermos definido no script do Streamlit.

````python
with button_col_3:
    if streamlit.button("Reiniciar", type="secondary", width="stretch"):
        if streamlit.session_state.get("contador", None) is not None:
            streamlit.session_state.clear()
````

#### Forçando Re-execução

Quando atualizamos um valor do session_state, a aplicação não re-renderiza os elementos. Então mesmo que o estado mude, não veremos o contador sempre ser atualizado quando clicarmos em um botão. Para forçar essa reatividade, podemos simplesmente re-executar todo o script após atualizarmos o session_state. Para isso, podemos chamar o método rerun do Streamlit. Abaixo, segue o exemplo completo do contador.

````python
streamlit.markdown(f"**Contagem**: {streamlit.session_state.get("contador", 0)}")

button_col_1, button_col_2, button_col_3 = streamlit.columns(3)

with button_col_1:
    if streamlit.button("Incrementar", type="primary", width="stretch"):
        if streamlit.session_state.get("contador", None) is None:
            streamlit.session_state.contador = 1
        else:
            streamlit.session_state.contador += 1

        streamlit.rerun()

with button_col_2:
    if streamlit.button("Decrementar", type="primary", width="stretch"):
        if streamlit.session_state.get("contador", None) is None:
            streamlit.session_state.contador = -1
        else:
            streamlit.session_state.contador -= 1
            
        streamlit.rerun()

with button_col_3:
    if streamlit.button("Reiniciar", type="secondary", width="stretch"):
        if streamlit.session_state.get("contador", None) is not None:
            streamlit.session_state.pop("contador")
            
        streamlit.rerun()
````

Abaixo, calculamos e criamos um dataframe que contém apenas duas colunas: Quinzena e Média de Interações.

````python
from ga4_churn_predictor.config import DATA_PATH

dataframe_users = pandas.read_parquet(DATA_PATH)
mean_total_interacoes_quinzena_1 = (
    dataframe_users["total_interacoes_quinzena_1"].mean().round(2)
)
mean_total_interacoes_quinzena_2 = (
    dataframe_users["total_interacoes_quinzena_2"].mean().round(2)
)
mean_total_interacoes_quinzena_3 = (
    dataframe_users["total_interacoes_quinzena_3"].mean().round(2)
)
mean_total_interacoes_quinzena_4 = (
    dataframe_users["total_interacoes_quinzena_4"].mean().round(2)
)

dataframe = pandas.DataFrame(
    {
        "Quinzena": [1, 2, 3, 4],
        "Média de Interações": [
            mean_total_interacoes_quinzena_1,
            mean_total_interacoes_quinzena_2,
            mean_total_interacoes_quinzena_3,
            mean_total_interacoes_quinzena_4,
        ],
    }
)
````

Também pode receber x_label e y_label, que definem o nome que daremos a métrica de cada eixo quando a figura do gráfico for desenhada. Também pode receber o parâmetro width que recebe "content" para ocupar apenas a largura do conteúdo ou "stretch" para ocupar toda a largura disponível. O parâmetro color define a cor principal do gráfico. Podendo receber códigos RGB (ex.: rgb(100, 100, 100)), códigos hexadecimais (ex.: #ff40ff), algumas cores por nome como "red", "green", "blue", etc. ou o nome de uma coluna do dataframe. Quando o color recebe o nome de uma coluna, a cor será determinada de acordo com o valor (ex.: "Não" seria branco e "Sim" seria azul). Por fim, dentre os parâmetros mais relevantes, o método pode receber o parâmetro height que define a altura do elemento do gráfico em pixels.

## Armazenamento em Cache

### Cache de Dados

Imagine o seguinte problema: em algum ponto do script, precisamos coletar uma grande quantidade de dados e que esse processo é demorado e toma recursos de processamento. Seria muito útil podermos armazenar os resultados da coleta em cache. Dessa forma, quando o usuário precisasse de novo dos mesmos dados, não precisaríamos buscá-los do zero. O Streamlit disponibiliza o decorator cache_data que faz exatamente isso. Ele deve ser aplicado em uma função que retorne dados como dataframes, objetos JSON, etc. O decorator pode receber alguns parâmetros como: show_spinner; show_time; max_entries; ttl; scope.

O parâmetro show_spinner define se um spinner de carregamento deverá ser exibido enquanto a função carrega os dados. O parâmetro show_time define se o tempo de espera é exibido ou não enquanto a função carrega os dados. O parâmetro max_entries define quantos registros aquela função poderá manter em cache. Quando chamamos a função com um certo conjunto de parâmetros, a função guarda o resultado retornado em cache. Se chamarmos a função de novo com os mesmos parâmetros, é retornado o valor em cache de acordo com aquele conjunto. Se chamarmos novamente com parâmetros diferentes, outro registro será criado. No exemplo abaixo, definimos que será guardado apenas um registro. Dessa forma, o dataframe em cache só terá uma única ocorrência e quando a função for chamada com outros parâmetros, a primeira ocorrência some. O parâmetro ttl (Time to Live) define por quanto tempo em segundos o valor poderá ser guardado em cache. No exemplo abaixo, definimos que o valor será guardado em cache por no máximo uma hora.

````python
@streamlit.cache_data(
    show_spinner=False, 
    show_time=False, 
    max_entries=1, 
    ttl=3600
)
def load_data() -> pandas.DataFrame:
    time.sleep(2)  # Simulando latência do servidor

    return pandas.read_csv(PROCESSED_DATA_DIR / "x_test.csv")
````

Para recuperar os dados, basta chamarmos a função definida.

````python
users = load_data()

streamlit.dataframe(users.sample(n=10), hide_index=True)
````

#### Fragmentação

Para resolver esse problema, podemos definir uma função que chama a outra função de carregamento dos dados e exibe o conteúdo que depende dos dados. Essa função deverá utilizar o decorator fragment do Streamlit. Na prática, o decorator separa a execução desse fragmento da execução do restante do script. Mas, utilizar o decorator não basta. Temos de passar para ele um parâmetro parallel, que define se a execução daquele fragmento ocorrerá em paralelo. Se sim, o restante do script não é interrompido.

````python
@streamlit.fragment(parallel=True)
def render_users_sample():
    with streamlit.spinner("Carregando amostra...", show_time=True):
        users = load_data()

        streamlit.dataframe(users.sample(n=10), hide_index=True)


render_users_sample()

streamlit.caption("Versão do modelo: v1.0.0 | Último retreino: Outubro/2026")
````

**Carregamento dos dados com execução de fragmento em paralelo**

![Figura 52](assets/images/figura_52.png)

O decorator fragment pode receber um parâmetro key, que deve ser um identificador único do fragmento. Mas para quê ele serve? Bom, com ele, podemos fazer com que o script re-execute apenas aquele fragmento, identificando-o pela chave. Ao chamar o método rerun do Streamlit, podemos passar o parâmetro scope. O parâmetro scope pode receber "app" para recarregar toda a aplicação, "fragment" para recarregar o fragmento atual, ou a chave de um fragmento se quisermor recarregá-lo de fora. No exemplo abaixo, toda vez que clicarmos no botão de "Buscar nova amostra", o fragmento será re-executado, o load_data retornará um valor guardado em cache do dataframe e exibirá uma amostra dele.

````python
@streamlit.fragment(key="amostra_dataframe_usuarios", parallel=True)
def render_users_sample():
    with streamlit.spinner("Carregando amostra...", show_time=True):
        users = load_data()

        streamlit.dataframe(users.sample(n=10), hide_index=True)

        if streamlit.button("Buscar nova amostra", width="stretch", type="primary"):
            streamlit.rerun(scope="fragment")
````

**Fragmento com botão de re-execução**

![Figura 53](assets/images/figura_53.png)

Se quisermos recarregar de fora do fragmento, devemos passar a chave do fragmento como parâmetro scope para o método rerun. Para isso, não podemos usar uma condicional. Temos de usar o parâmetro on_click do botão, que recebe uma função que reage ao clique. Observando a imagem em tela, nada mudou na aparência. A única diferença é que agora o botão está fora do fragmento e por isso sempre é exibido, mesmo enquanto os dados carregam.

````python
@streamlit.fragment(key="amostra_dataframe_usuarios", parallel=True)
def render_users_sample():
    with streamlit.spinner("Carregando amostra...", show_time=True):
        users = load_data()

        streamlit.dataframe(users.sample(n=10), hide_index=True)


render_users_sample()

streamlit.button(
    "Buscar nova amostra",
    width="stretch",
    type="primary",
    on_click=lambda: streamlit.rerun(scope="amostra_dataframe_usuarios"),
)
````

**Botão de recarregamento externo ao fragmento**

![Figura 54](assets/images/figura_54.png)

O uso do método rerun com scope não funciona com botões em condicional, como no exemplo abaixo.

````python
if streamlit.button(
    "Buscar nova amostra",
    width="stretch",
    type="primary",
):
    streamlit.rerun(scope="amostra_dataframe_usuarios")
````
